```python
from flask import Flask, request, jsonify
from flask_cors import CORS
from datetime import datetime
import uuid
import json
import os

app = Flask(__name__)
CORS(app)

DATA_FILE = "agent_framework/data/messages.json"

agents = {
    "Agent-A": {
        "name": "Agent-A",
        "role": "Request Agent",
        "status": "Online"
    },
    "Agent-B": {
        "name": "Agent-B",
        "role": "Processing Agent",
        "status": "Online"
    }
}

if not os.path.exists(DATA_FILE):
    with open(DATA_FILE, "w") as f:
        json.dump([], f)


def load_messages():
    with open(DATA_FILE, "r") as f:
        return json.load(f)


def save_messages(messages):
    with open(DATA_FILE, "w") as f:
        json.dump(messages, f, indent=4)


@app.route("/")
def home():
    return jsonify({
        "framework": "Agent Communication Framework",
        "status": "Running",
        "message": "Agent communication server is active"
    })


@app.route("/api/agents", methods=["GET"])
def get_agents():
    return jsonify(list(agents.values()))


@app.route("/api/send", methods=["POST"])
def send_request():

    data = request.get_json()

    sender = data.get("sender")
    receiver = data.get("receiver")
    message = data.get("message")

    if not sender or not receiver or not message:
        return jsonify({
            "success": False,
            "error": "Sender, receiver and message are required"
        }), 400

    if sender not in agents or receiver not in agents:
        return jsonify({
            "success": False,
            "error": "Unknown agent"
        }), 400

    request_id = str(uuid.uuid4())[:8]

    request_message = {
        "id": request_id,
        "type": "REQUEST",
        "sender": sender,
        "receiver": receiver,
        "message": message,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "status": "Sent"
    }

    messages = load_messages()
    messages.append(request_message)
    save_messages(messages)

    return jsonify({
        "success": True,
        "request_id": request_id,
        "message": request_message
    })


@app.route("/api/respond", methods=["POST"])
def send_response():

    data = request.get_json()

    request_id = data.get("request_id")
    sender = data.get("sender")
    receiver = data.get("receiver")
    message = data.get("message")

    if not request_id or not sender or not receiver or not message:
        return jsonify({
            "success": False,
            "error": "Request ID, sender, receiver and message are required"
        }), 400

    response_message = {
        "id": str(uuid.uuid4())[:8],
        "request_id": request_id,
        "type": "RESPONSE",
        "sender": sender,
        "receiver": receiver,
        "message": message,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "status": "Delivered"
    }

    messages = load_messages()
    messages.append(response_message)
    save_messages(messages)

    return jsonify({
        "success": True,
        "message": response_message
    })


@app.route("/api/messages", methods=["GET"])
def get_messages():

    messages = load_messages()

    return jsonify({
        "success": True,
        "count": len(messages),
        "messages": messages
    })


@app.route("/api/clear", methods=["DELETE"])
def clear_messages():

    save_messages([])

    return jsonify({
        "success": True,
        "message": "Communication log cleared"
    })


@app.route("/api/health", methods=["GET"])
def health():

    return jsonify({
        "status": "Online",
        "framework": "Agent Communication Framework",
        "agents": len(agents),
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    })


print("Agent Communication Framework loaded successfully!")
print("Available API endpoints:")
print("/api/agents")
print("/api/send")
print("/api/respond")
print("/api/messages")
print("/api/clear")
print("/api/health")
```

```python
def process_agent_request(request_id, sender, receiver, message):

    if receiver != "Agent-B":
        return {
            "success": False,
            "error": "Agent-B is the configured processing agent"
        }

    message_lower = message.lower()

    if "temperature" in message_lower:
        response_text = (
            "Agent-B received the temperature request and "
            "processed the information successfully."
        )

    elif "humidity" in message_lower:
        response_text = (
            "Agent-B received the humidity request and "
            "processed the information successfully."
        )

    elif "status" in message_lower:
        response_text = (
            "Agent-B is online and ready to process requests."
        )

    elif "hello" in message_lower or "hi" in message_lower:
        response_text = (
            "Hello Agent-A! Agent-B received your message "
            "and is ready for communication."
        )

    else:
        response_text = (
            "Agent-B received the request: '" +
            message +
            "'. The request has been processed successfully."
        )

    response_data = {
        "request_id": request_id,
        "sender": receiver,
        "receiver": sender,
        "message": response_text
    }

    return send_agent_response(response_data)


def send_agent_response(data):

    request_id = data["request_id"]
    sender = data["sender"]
    receiver = data["receiver"]
    message = data["message"]

    response_message = {
        "id": str(uuid.uuid4())[:8],
        "request_id": request_id,
        "type": "RESPONSE",
        "sender": sender,
        "receiver": receiver,
        "message": message,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "status": "Delivered"
    }

    messages = load_messages()
    messages.append(response_message)
    save_messages(messages)

    return response_message


print("Agent processing module created successfully!")
print("Agent-B is ready to process requests.")
```

```python
# Replace the existing /api/send route with an automatic request-response route

@app.route("/api/send_auto", methods=["POST"])
def send_auto_request():

    data = request.get_json()

    sender = data.get("sender")
    receiver = data.get("receiver")
    message = data.get("message")

    if not sender or not receiver or not message:
        return jsonify({
            "success": False,
            "error": "Sender, receiver and message are required"
        }), 400

    if sender not in agents or receiver not in agents:
        return jsonify({
            "success": False,
            "error": "Unknown agent"
        }), 400

    request_id = str(uuid.uuid4())[:8]

    request_message = {
        "id": request_id,
        "type": "REQUEST",
        "sender": sender,
        "receiver": receiver,
        "message": message,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "status": "Sent"
    }

    messages = load_messages()
    messages.append(request_message)
    save_messages(messages)

    # Agent-B processes the request automatically
    response_message = process_agent_request(
        request_id,
        sender,
        receiver,
        message
    )

    return jsonify({
        "success": True,
        "request": request_message,
        "response": response_message
    })


print("Automatic request-response API created!")
print("Endpoint: /api/send_auto")
```

```python
# ================================
# FINAL WORKING VERSION
# ================================

from flask import Flask, request, jsonify, Response
from flask_cors import CORS
from datetime import datetime
import uuid
import threading
import time
from pyngrok import ngrok

app_final = Flask(__name__)
CORS(app_final)

messages_final = []

agents_final = {
    "Agent-A": {
        "name": "Agent-A",
        "role": "Request Agent",
        "status": "Online"
    },
    "Agent-B": {
        "name": "Agent-B",
        "role": "Processing Agent",
        "status": "Online"
    }
}

# --------------------------------
# WEBSITE
# --------------------------------

@app_final.route("/")
@app_final.route("/dashboard")
def dashboard_final():

    page = """
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">

<title>Agent Communication Framework</title>

<style>

* {
    box-sizing: border-box;
}

body {
    margin: 0;
    font-family: Arial, sans-serif;
    background: #eef2f7;
    color: #172033;
}

.header {
    background: #111827;
    color: white;
    padding: 25px 35px;
}

.header h1 {
    margin: 0;
    font-size: 28px;
}

.header p {
    margin: 7px 0 0;
    color: #cbd5e1;
}

.online {
    float: right;
    color: #4ade80;
    font-weight: bold;
}

.container {
    padding: 25px;
    max-width: 1400px;
    margin: auto;
}

.agents {
    display: grid;
    grid-template-columns: 1fr 1fr 1fr;
    gap: 20px;
    margin-bottom: 20px;
}

.card {
    background: white;
    border-radius: 15px;
    padding: 22px;
    box-shadow: 0 4px 15px rgba(0,0,0,.08);
}

.agent {
    display: flex;
    align-items: center;
    gap: 15px;
}

.icon {
    width: 55px;
    height: 55px;
    background: #2563eb;
    color: white;
    border-radius: 50%;
    display: flex;
    align-items: center;
    justify-content: center;
    font-weight: bold;
    font-size: 20px;
}

.agent h3 {
    margin: 0 0 5px;
}

.status {
    color: #16a34a;
    font-weight: bold;
    font-size: 13px;
}

.number {
    font-size: 32px;
    font-weight: bold;
}

.main {
    display: grid;
    grid-template-columns: 380px 1fr;
    gap: 20px;
}

label {
    display: block;
    font-weight: bold;
    margin-top: 15px;
    margin-bottom: 6px;
}

select,
textarea {
    width: 100%;
    padding: 12px;
    border: 1px solid #cbd5e1;
    border-radius: 8px;
    font-size: 15px;
}

textarea {
    resize: vertical;
}

button {
    width: 100%;
    margin-top: 18px;
    padding: 13px;
    border: 0;
    border-radius: 8px;
    background: #2563eb;
    color: white;
    font-weight: bold;
    font-size: 15px;
    cursor: pointer;
}

button:hover {
    background: #1d4ed8;
}

.clear {
    background: #dc2626;
    width: auto;
    float: right;
    margin-top: 0;
}

.flow {
    margin-top: 25px;
    padding: 18px;
    background: #f8fafc;
    border-radius: 10px;
    text-align: center;
}

.flowbox {
    display: inline-block;
    padding: 10px 15px;
    background: white;
    border: 2px solid #2563eb;
    border-radius: 8px;
    font-weight: bold;
    margin: 5px;
}

.log {
    height: 430px;
    overflow-y: auto;
    margin-top: 20px;
}

.message {
    padding: 15px;
    margin-bottom: 12px;
    border-radius: 10px;
}

.request {
    background: #dbeafe;
    border-left: 5px solid #2563eb;
}

.response {
    background: #dcfce7;
    border-left: 5px solid #16a34a;
}

.type {
    font-size: 12px;
    font-weight: bold;
}

.messageText {
    margin-top: 6px;
    font-size: 16px;
}

.meta {
    margin-top: 7px;
    color: #64748b;
    font-size: 11px;
}

.success {
    color: #16a34a;
    font-weight: bold;
    margin-top: 12px;
}

.empty {
    text-align: center;
    color: #94a3b8;
    padding: 100px 20px;
}

@media(max-width:900px) {
    .agents,
    .main {
        grid-template-columns: 1fr;
    }
}

</style>
</head>

<body>

<div class="header">

<div class="online">● SYSTEM ONLINE</div>

<h1>Agent Communication Framework</h1>

<p>Multi-Agent Request and Response Communication System</p>

</div>


<div class="container">


<div class="agents">

<div class="card agent">

<div class="icon">A</div>

<div>

<h3>Agent-A</h3>

<div>Request Agent</div>

<div class="status">● Online</div>

</div>

</div>


<div class="card agent">

<div class="icon">B</div>

<div>

<h3>Agent-B</h3>

<div>Processing Agent</div>

<div class="status">● Online</div>

</div>

</div>


<div class="card">

<h3>Communication Statistics</h3>

<div style="display:flex;gap:50px">

<div>

<div class="number" id="requests">0</div>

<div>Requests</div>

</div>

<div>

<div class="number" id="responses">0</div>

<div>Responses</div>

</div>

</div>

</div>

</div>


<div class="main">


<div class="card">

<h2>Send Agent Request</h2>

<label>Sender</label>

<select id="sender">

<option value="Agent-A">Agent-A</option>

<option value="Agent-B">Agent-B</option>

</select>


<label>Receiver</label>

<select id="receiver">

<option value="Agent-B">Agent-B</option>

<option value="Agent-A">Agent-A</option>

</select>


<label>Request Message</label>

<textarea
id="message"
rows="5"
placeholder="Enter request message..."
></textarea>


<button onclick="sendRequest()">
SEND REQUEST
</button>


<div id="result"></div>


<div class="flow">

<b>COMMUNICATION FLOW</b>

<br><br>

<span class="flowbox">Agent-A</span>
→
<span class="flowbox">Framework</span>
→
<span class="flowbox">Agent-B</span>

<br>

↓

<br>

<span class="flowbox">Response</span>

</div>

</div>


<div class="card">

<button class="clear" onclick="clearLog()">
CLEAR LOG
</button>

<h2>Communication Log</h2>

<div id="log" class="log">

<div class="empty">
No communication messages yet.<br>
Send a request to start communication.
</div>

</div>

</div>

</div>

</div>


<script>

async function sendRequest() {

    const sender =
        document.getElementById("sender").value;

    const receiver =
        document.getElementById("receiver").value;

    const message =
        document.getElementById("message").value.trim();


    if (!message) {

        alert("Please enter a message.");

        return;

    }


    const result =
        document.getElementById("result");


    result.innerHTML =
        "<div style='color:#2563eb;font-weight:bold'>Processing request...</div>";


    try {

        const response = await fetch(
            "/api/send",
            {
                method: "POST",

                headers: {
                    "Content-Type": "application/json"
                },

                body: JSON.stringify({
                    sender: sender,
                    receiver: receiver,
                    message: message
                })
            }
        );


        const data = await response.json();


        if (data.success) {

            document.getElementById("message").value = "";


            result.innerHTML =
                "<div class='success'>✓ Request sent and response received!</div>";


            // DIRECTLY ADD THE RESPONSE TO THE SCREEN
            addMessage(data.request);
            addMessage(data.response);


            updateCounts();

        }

        else {

            result.innerHTML =
                "<div style='color:red'>Error: " +
                data.error +
                "</div>";

        }

    }

    catch(error) {

        result.innerHTML =
            "<div style='color:red'>Connection error</div>";

        console.log(error);

    }

}


function addMessage(msg) {

    if (!msg) return;


    const log =
        document.getElementById("log");


    if (log.innerHTML.includes("No communication")) {

        log.innerHTML = "";

    }


    const div =
        document.createElement("div");


    div.className =
        "message " +
        (msg.type === "REQUEST"
            ? "request"
            : "response");


    div.innerHTML =

        "<div class='type'>" +
        msg.type +
        "</div>" +

        "<strong>" +
        msg.sender +
        " → " +
        msg.receiver +
        "</strong>" +

        "<div class='messageText'>" +
        msg.message +
        "</div>" +

        "<div class='meta'>" +
        "ID: " +
        msg.id +
        " | " +
        msg.timestamp +
        " | " +
        msg.status +
        "</div>";


    log.prepend(div);

}


function updateCounts() {

    const all =
        document.querySelectorAll(".message");


    let req = 0;
    let res = 0;


    all.forEach(function(x) {

        if (x.classList.contains("request"))
            req++;

        if (x.classList.contains("response"))
            res++;

    });


    document.getElementById("requests").innerText = req;

    document.getElementById("responses").innerText = res;

}


async function clearLog() {

    await fetch(
        "/api/clear",
        {
            method: "DELETE"
        }
    );


    document.getElementById("log").innerHTML =
        "<div class='empty'>No communication messages yet.<br>Send a request to start communication.</div>";


    document.getElementById("requests").innerText = "0";

    document.getElementById("responses").innerText = "0";

}


</script>

</body>
</html>
"""

    return Response(page, mimetype="text/html")


# --------------------------------
# SEND REQUEST
# --------------------------------

@app_final.route("/api/send", methods=["POST"])
def send_final():

    data = request.get_json()

    sender = data.get("sender")
    receiver = data.get("receiver")
    message = data.get("message")

    if not sender or not receiver or not message:

        return jsonify({
            "success": False,
            "error": "Missing information"
        }), 400


    request_id = str(uuid.uuid4())[:8]


    req = {
        "id": request_id,
        "type": "REQUEST",
        "sender": sender,
        "receiver": receiver,
        "message": message,
        "timestamp": datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        ),
        "status": "Sent"
    }


    messages_final.append(req)


    # Agent-B response
    response_text = (
        "Agent-B received the request '" +
        message +
        "' and processed it successfully."
    )


    if "hello" in message.lower():

        response_text = (
            "Hello Agent-A! Agent-B received your "
            "message successfully."
        )


    response = {
        "id": str(uuid.uuid4())[:8],
        "request_id": request_id,
        "type": "RESPONSE",
        "sender": receiver,
        "receiver": sender,
        "message": response_text,
        "timestamp": datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        ),
        "status": "Delivered"
    }


    messages_final.append(response)


    return jsonify({
        "success": True,
        "request": req,
        "response": response
    })


# --------------------------------
# GET MESSAGES
# --------------------------------

@app_final.route("/api/messages")
def messages_api():

    return jsonify({
        "success": True,
        "messages": messages_final
    })


# --------------------------------
# CLEAR
# --------------------------------

@app_final.route("/api/clear", methods=["DELETE"])
def clear_api():

    messages_final.clear()

    return jsonify({
        "success": True
    })


# --------------------------------
# START SERVER
# --------------------------------

def run_final():

    app_final.run(
        host="0.0.0.0",
        port=5002,
        debug=False,
        use_reloader=False
    )


ngrok.kill()

threading.Thread(
    target=run_final,
    daemon=True
).start()

time.sleep(3)

final_url = ngrok.connect(5002)

print("=" * 55)
print("FINAL WEBSITE IS READY")
print("=" * 55)
print()
print("OPEN THIS URL:")
print(str(final_url))
print()
print("DASHBOARD:")
print(str(final_url) + "/dashboard")
print()
print("=" * 55)
```
